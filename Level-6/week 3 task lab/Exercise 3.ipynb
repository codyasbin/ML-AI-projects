```python
import time  

# Define the graph for the first part  
graph = {  
    'A': ['B', 'C'],  
    'B': ['A', 'E', 'D'],  
    'C': ['A', 'F'],  
    'D': ['B', 'G'],  
    'E': ['B', 'F'],  
    'F': ['C', 'E', 'H'],  
    'G': ['D', 'I'],  
    'H': ['F'],  
    'I': ['G', 'J'],  
    'J': ['I', 'K'],  
    'K': ['J']  
}  

# Define the DLS function  
def dls(graph, node, goal, depth, visited=None):  
    if visited is None:  
        visited = set()  
    if depth == 0:  
        return None  
    if node == goal:  
        return [node]  
    if node not in visited:  
        visited.add(node)  
        for neighbor in graph[node]:  
            path = dls(graph, neighbor, goal, depth - 1, visited)  
            if path:  
                return [node] + path  
    return None  

# Example usage with different max depths  
start_node = 'A'  
goal_node = 'H'  

print("Depth-Limited Search (DLS):")  
for max_depth in range(1, 6):  
    result = dls(graph, start_node, goal_node, max_depth)  
    if result:  
        print(f"Max Depth {max_depth}: Path found: {result}")  
    else:  
        print(f"Max Depth {max_depth}: No path found.")  

# Define a more complex graph for the comparison part  
complex_graph = {  
    'A': ['B', 'C'],  
    'B': ['D', 'E'],  
    'C': ['F', 'G'],  
    'D': ['H', 'I'],  
    'E': ['J'],  
    'F': ['K', 'L'],  
    'G': ['M'],  
    'H': ['N'],  
    'I': ['O'],  
    'J': ['P'],  
    'K': ['Q'],  
    'L': ['R'],  
    'M': ['S'],  
    'N': [],  
    'O': [],  
    'P': [],  
    'Q': [],  
    'R': [],  
    'S': []  
}  

# Define the DFS function  
def dfs(graph, node, goal, visited=None):  
    if visited is None:  
        visited = set()  
    if node == goal:  
        return [node]  
    if node not in visited:  
        visited.add(node)  
        for neighbor in graph[node]:  
            path = dfs(graph, neighbor, goal, visited)  
            if path:  
                return [node] + path  
    return None  

# Define the BFS function  
def bfs(graph, start, goal):  
    visited = set()  
    queue = [(start, [start])]  

    while queue:  
        current, path = queue.pop(0)  
        if current == goal:  
            return path  
        if current not in visited:  
            visited.add(current)  
            for neighbor in graph[current]:  
                queue.append((neighbor, path + [neighbor]))  

    return None  

# Comparison of DLS, BFS, DFS on the complex graph  
start_node = 'A'  
goal_node = 'S'  

# Test DLS with different depths  
print("\nDepth-Limited Search (DLS):")  
for max_depth in range(1, 6):  
    start_time = time.time()  
    result = dls(complex_graph, start_node, goal_node, max_depth)  
    end_time = time.time()  
    if result:  
        print(f"Max Depth {max_depth}: Path found: {result} (Time taken: {end_time - start_time:.6f} seconds)")  
    else:  
        print(f"Max Depth {max_depth}: No path found. (Time taken: {end_time - start_time:.6f} seconds)")  

# Test BFS  
print("\nBreadth-First Search (BFS):")  
start_time = time.time()  
bfs_result = bfs(complex_graph, start_node, goal_node)  
end_time = time.time()  
if bfs_result:  
    print(f"BFS: Path found: {bfs_result} (Time taken: {end_time - start_time:.6f} seconds)")  
else:  
    print("BFS: No path found.")  

# Test DFS  
print("\nDepth-First Search (DFS):")  
start_time = time.time()  
dfs_result = dfs(complex_graph, start_node, goal_node)  
end_time = time.time()  
if dfs_result:  
    print(f"DFS: Path found: {dfs_result} (Time taken: {end_time - start_time:.6f} seconds)")  
else:  
    print("DFS: No path found.")
```

# In this exercise we will be using search algorithms to solve a maze problem. 


```python
from collections import deque  

def bfs_maze(maze, start, goal):  
    rows, cols = len(maze), len(maze[0])  
    visited = set()  
    parent = {}  # To keep track of the parent of each cell in the path  
    
    def is_valid(x, y):  
        return 0 <= x < rows and 0 <= y < cols and maze[x][y] != '#' and (x, y) not in visited  

    queue = deque([start])  # Initialize the queue with the starting position  
    visited.add(start)  # Mark the starting position as visited  
    found = False  
    
    while queue:  
        x, y = queue.popleft()  

        if (x, y) == goal:  # Goal found  
            found = True  
            break  

        # Define possible moves: right, down, left, up  
        moves = [(0, 1), (1, 0), (0, -1), (-1, 0)]  
        for dx, dy in moves:  
            new_x, new_y = x + dx, y + dy  
            if is_valid(new_x, new_y):  
                queue.append((new_x, new_y))  
                visited.add((new_x, new_y))  
                parent[(new_x, new_y)] = (x, y)  

    if not found:  
        return "No path found"  

    # Reconstruct the path from the goal to the start  
    path = []  
    x, y = goal  
    while (x, y) != start:  
        path.append((x, y))  
        x, y = parent[(x, y)]  
    path.append(start)  
    path.reverse()  # Reverse the path to go from start to goal  

    # Update the maze with the path  
    for x, y in path:  
        if maze[x][y] != 'S' and maze[x][y] != 'G':  
            maze[x][y] = '*'  

    return "Path found"  

# Define the corrected maze  
maze = [  
    ['S', '.', '.', '.', '.', '.', '#', '.', '.'],  
    ['#', '#', '#', '#', '#', '.', '#', '#', '#'],  
    ['.', '.', '.', '.', '.', '.', '.', '.', 'G'],  
    ['#', '#', '#', '#', '#', '#', '#', '#', '#']  
]  

# Starting and goal positions  
start_position = (0, 0)  
goal_position = (2, 8)  

result = bfs_maze(maze, start_position, goal_position)  

# Print the maze with the path  
for row in maze:  
    print(' '.join(row))  
print(result)
```
