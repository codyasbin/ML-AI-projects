```python
import matplotlib.pyplot as plt  
import networkx as nx  

# Define the graph based on the provided diagram  
graph = {  
    'A': ['B', 'C'],  
    'B': ['A', 'D', 'E'],  
    'C': ['A', 'F'],  
    'D': ['B'],  
    'E': ['B', 'F'],  
    'F': ['C', 'E'],  
    'G': ['D', 'H'],  
    'H': ['G', 'F']  
}  

# Create a directed graph from the adjacency list  
G = nx.Graph(graph)  

# Set positions for the nodes  
pos = nx.spring_layout(G)  

# Draw the graph  
nx.draw(G, pos, with_labels=True, node_size=1000, node_color='lightblue', font_size=15, font_weight='bold')  

# Title and display the graph  
plt.title("Graph Visualization")  
plt.show()
```

```python
from collections import deque  

def bfs(graph, start, goal):  
    visited = set()  
    queue = deque([[start, []]])  

    while queue:  
        node, path = queue.popleft()  
        if node == goal:  
            return path + [node]  
        if node not in visited:  
            visited.add(node)  
            for neighbor in graph[node]:  
                queue.append((neighbor, path + [node]))  

    return None  # Goal not found  

graph = {  
    'A': ['B', 'C'],  
    'B': ['A', 'D', 'E'],  
    'C': ['A', 'F'],  
    'D': ['B'],  
    'E': ['B', 'F'],  
    'F': ['C', 'E']  
}
start_node = 'A'  
goal_node = 'E'  
result = bfs(graph, start_node, goal_node)  

if result:  
    print("Path found:", result)  
else:  
    print("No path found.")
```

```python
from collections import deque  
import time  
import matplotlib.pyplot as plt  
import networkx as nx  

def bfs(graph, start, goal):  
    visited = set()  
    queue = deque([[start, []]])  

    while queue:  
        node, path = queue.popleft()  
        if node == goal:  
            return path + [node]  
        if node not in visited:  
            visited.add(node)  
            for neighbor in graph[node]:  
                queue.append((neighbor, path + [node]))  

    return None  # Goal not found  

# Define the new graph based on the provided diagram  
graph = {  
    'A': ['B', 'C'],  
    'B': ['A', 'E', 'D'],  
    'C': ['A', 'F'],  
    'D': ['B', 'G'],  
    'E': ['B', 'F'],  
    'F': ['C', 'E', 'H'],  
    'G': ['D'],  
    'H': ['F']  
}  

# Visualize the graph  
def visualize_graph(graph):  
    G = nx.Graph(graph)  
    pos = nx.spring_layout(G)  
    nx.draw(G, pos, with_labels=True, node_size=1000, node_color='lightblue', font_size=15, font_weight='bold')  
    plt.title("Graph Visualization")  
    plt.show()  

# Call the visualization function  
visualize_graph(graph)  

# Change starting and ending points  
start_points = ['A', 'B', 'C']  
goal_points = ['F', 'D', 'H']  

# Measure time for different start and end points  
for start, goal in zip(start_points, goal_points):  
    start_time = time.time()  
    result = bfs(graph, start, goal)  
    end_time = time.time()  
    
    if result:  
        print(f"Path from {start} to {goal}: {result} (Time taken: {end_time - start_time:.6f} seconds)")  
    else:  
        print(f"No path found from {start} to {goal} (Time taken: {end_time - start_time:.6f} seconds)")
```
