```python
# Define the graph  
graph = {  
    'A': ['B', 'C'],  
    'B': ['A', 'E', 'D'],  
    'C': ['A', 'F'],  
    'D': ['B', 'G'],  
    'E': ['B', 'F'],  
    'F': ['C', 'E', 'H'],  
    'G': ['D'],  
    'H': ['F']  
}  

# Define the DFS function  
def dfs(graph, node, goal, visited=None):  
    if visited is None:  
        visited = set()  
    if node == goal:  
        return [node]  
    if node not in visited:  
        visited.add(node)  
        for neighbor in graph[node]:  
            path = dfs(graph, neighbor, goal, visited)  
            if path:  
                return [node] + path  
    return None  # Goal not found  

# Example usage  
start_node = 'A'  # Specify the starting node  
goal_node = 'H'   # Specify the goal node  

result = dfs(graph, start_node, goal_node)  
if result:  
    print("Path found:", result)  
else:  
    print("No path found.")
```

# Q: Use the time library to calculate how long would it take to solve the problem using three different starting points and ending points

```python
import time  

# Define the graph  
graph = {  
    'A': ['B', 'C'],  
    'B': ['A', 'E', 'D'],  
    'C': ['A', 'F'],  
    'D': ['B', 'G'],  
    'E': ['B', 'F'],  
    'F': ['C', 'E', 'H'],  
    'G': ['D'],  
    'H': ['F']  
}  

# Define the DFS function  
def dfs(graph, node, goal, visited=None):  
    if visited is None:  
        visited = set()  
    if node == goal:  
        return [node]  
    if node not in visited:  
        visited.add(node)  
        for neighbor in graph[node]:  
            path = dfs(graph, neighbor, goal, visited)  
            if path:  
                return [node] + path  
    return None  # Goal not found  

# Example usage with timing  
start_goal_pairs = [('A', 'H'), ('B', 'F'), ('C', 'D')]  

for start, goal in start_goal_pairs:  
    start_time = time.time()  # Start time  
    result = dfs(graph, start, goal)  
    end_time = time.time()  # End time  
    
    if result:  
        print(f"Path from {start} to {goal}: {result} (Time taken: {end_time - start_time:.6f} seconds)")  
    else:  
        print(f"No path found from {start} to {goal} (Time taken: {end_time - start_time:.6f} seconds).")
```
