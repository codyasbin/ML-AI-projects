```python
import cv2
import numpy as np
import time

def detect_numbers(frame):
    # Convert to HSV color space
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Define a tighter color range for high intensity red
    lower_red = np.array([160, 150, 150])  # Adjust as needed
    upper_red = np.array([179, 255, 255])

    # Create a mask for the specified red color range
    mask = cv2.inRange(hsv, lower_red, upper_red)

    # Apply morphological operations to remove noise
    kernel = np.ones((5, 5), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)

    # Find contours of potential numbers in the masked image
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # Check if any contour has a rectangular shape (indicative of a number)
    for cnt in contours:
        approx = cv2.approxPolyDP(cnt, 0.02 * cv2.arcLength(cnt, True), True)
        if len(approx) == 4:  # Approximately rectangular
            return True

    return False

# Initialize camera
cap = cv2.VideoCapture(0)

# Initialize a flag to track the previous detection state
previous_detection = False
detection_timer = 0  # Timer to track detection duration

while True:
    # Capture frame-by-frame
    ret, frame = cap.read()

    # Detect numbers in the frame
    current_detection = detect_numbers(frame)

    # Check if the detection state has changed
    if current_detection and not previous_detection:
        detection_timer = 30  # Set timer for 30 seconds
        print("Vehicle Detected!")
    elif current_detection and previous_detection:
        detection_timer -= 1
        if detection_timer <= 0:
            print("Vehicle not Detected.")
            detection_timer = 0

    previous_detection = current_detection

    # Display the frame
    cv2.imshow('Frame', frame)

    # Exit if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release the capture and close windows
cap.release()
cv2.destroyAllWindows()
```
